```python
import json
import ollama
from pydantic import BaseModel

# 1. Define the desired output schema
class ExtractedData(BaseModel):
    customer_name: str
    phone_number: str
    order_number: str
    delivery_address: str

# 2. Raw unstructured text you want to process
sample_text = """
Hi, my name is John Smith. Please ship my order #98765 to 123 Main St, 
Austin, TX. You can reach me at 555-0199 if there are any issues.
"""

# 3. Get JSON schema from Pydantic
schema = ExtractedData.model_json_schema()

# 4. Call Llama 3.2 with structured format enforcement
response = ollama.chat(
    model='llama3.2',
    messages=[
        {'role': 'system', 'content': 'You are a precise data extraction assistant. Extract information strictly matching the requested schema.'},
        {'role': 'user', 'content': f'Extract information from this text:\n\n{sample_text}'}
    ],
    format=schema, # Enforces JSON matching the Pydantic model
    options={'temperature': 0}
)

# 5. Parse and display the result
extracted_json = json.loads(response['message']['content'])
print(json.dumps(extracted_json, indent=2))
```
